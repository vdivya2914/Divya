```python
import os
import shutil

print("===== FILE CLEANUP TOOL =====")

# Get folder path from user
folder_path = input("Enter folder path: ")

# Check whether folder exists
if not os.path.exists(folder_path):
    print("Folder does not exist!")
else:
    # Create a separate folder for unwanted files
    cleanup_folder = os.path.join(folder_path, "Cleaned_Files")

    if not os.path.exists(cleanup_folder):
        os.mkdir(cleanup_folder)

    # Unwanted file extensions
    unwanted_extensions = [".tmp", ".log", ".bak", ".temp"]

    count = 0

    # Scan files in the folder
    for file in os.listdir(folder_path):

        file_path = os.path.join(folder_path, file)

        # Check only files
        if os.path.isfile(file_path):

            # Get file extension
            extension = os.path.splitext(file)[1].lower()

            # Check unwanted files
            if extension in unwanted_extensions:

                destination = os.path.join(cleanup_folder, file)

                shutil.move(file_path, destination)

                print("Moved:", file)
                count += 1

    print("\n===== CLEANUP COMPLETED =====")
    print("Files moved:", count)
    print("Cleaned files are stored in:", cleanup_folder)
```
